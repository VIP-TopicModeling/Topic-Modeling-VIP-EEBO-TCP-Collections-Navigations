```python
# Importing the necessary libraries
import numpy as np
import pandas as pd

```

```python
# BIBLIOGRAPHICAL Master Data
biblio_data = pd.read_csv("../master_biblio_data.csv")
# biblio_data.head()
biblio_data.info()
biblio_data
```

```python
# Loop over FIles in Navigations_headed_xml A0-A5
import os
folders = ["A0-A5", "A6-9", "B"]
eebo_tcp_consolidation = pd.DataFrame(columns=biblio_data.columns)
eebo_tcp_consolidation["XML_File_Path"] = []
no_rows_not_found = 0
for folder in folders:
    for file in os.listdir("../Navigations_headed_xml/" + folder):
        # print(file)
        tcp_no = file.split('.')[0]
        biblio_row = biblio_data[biblio_data['tcpNO'] == tcp_no]
        if not biblio_row.empty:
            # print(biblio_row)
            biblio_row_copy = biblio_row.copy()
            biblio_row_copy["XML_File_Path"] = "https://github.com/VIP-TopicModeling/Topic-Modeling-VIP-EEBO-TCP-Collections-Navigations/tree/master/Navigations_headed_xml/" + folder + "/" + file
            # print(file_path)
            eebo_tcp_consolidation = pd.concat([eebo_tcp_consolidation, biblio_row_copy], ignore_index=True)
        else:
            no_rows_not_found += 1
            print(f"No row found for {tcp_no}")

eebo_tcp_consolidation.to_csv("eebo_tcp_consolidation.csv", index=False)
```
